## Task 1: Basic Class & Object Creation

```python
# Task 1: Basic Class & Object Creation
# Create a Product class with attributes: name, price, category
# Method: get_info() - prints product details

class Product:
    """Product class for e-commerce store"""
    
    def __init__(self, name, price, category):
        """Initialize product attributes"""
        self.name = name
        self.price = price
        self.category = category
    
    def get_info(self):
        """Print product information"""
        print(f"Product: {self.name}")
        print(f"Price: ${self.price}")
        print(f"Category: {self.category}")
    
    def apply_discount(self, percent):
        """Apply discount and return discounted price"""
        discount_amount = self.price * (percent / 100)
        discounted_price = self.price - discount_amount
        return discounted_price

# Create two product objects
print("Creating Product Objects:")
print("=" * 40)

product1 = Product("Laptop", 1200, "Electronics")
product2 = Product("T-Shirt", 25, "Clothing")

# Test get_info() method
print("\nProduct 1 Info:")
product1.get_info()

print("\nProduct 2 Info:")
product2.get_info()

# Test apply_discount() method
print("\n" + "=" * 40)
print("Testing apply_discount():")
print("=" * 40)

original_price = product1.price
discounted_price = product1.apply_discount(10)
print(f"Product: {product1.name}")
print(f"Original Price: ${original_price}")
print(f"Discount: 10%")
print(f"Discounted Price: ${discounted_price}")
```

## Task 2: Constructor & Encapsulation

```python
# Task 2: Constructor & Encapsulation
# Modify Product class with private attribute _price
# Create getter and setter methods

class ProductV2:
    """Product class with encapsulation"""
    
    def __init__(self, name, price, category):
        """Initialize product with private price attribute"""
        self.name = name
        self._price = price  # Private attribute
        self.category = category
    
    def get_info(self):
        """Print product information"""
        print(f"Product: {self.name}")
        print(f"Price: ${self._price}")
        print(f"Category: {self.category}")
    
    # Getter method
    def get_price(self):
        """Get the product price"""
        return self._price
    
    # Setter method
    def set_price(self, new_price):
        """Set the product price (only if new_price > 0)"""
        if new_price > 0:
            self._price = new_price
            print(f"Price updated to ${new_price}")
        else:
            print("Error: Price must be greater than 0")
    
    def apply_discount(self, percent):
        """Apply discount to current price"""
        discount_amount = self._price * (percent / 100)
        return self._price - discount_amount

# Test encapsulation
print("Testing Encapsulation:")
print("=" * 40)

product = ProductV2("iPhone", 999, "Electronics")

print("\nOriginal Product Info:")
product.get_info()

print("\n" + "=" * 40)
print("Using Getter and Setter:")
print("=" * 40)

# Use getter
print(f"\nGetting price: ${product.get_price()}")

# Use setter with valid price
print("\nSetting new price to $850:")
product.set_price(850)

# Try to set invalid price
print("\nTrying to set invalid price (-100):")
product.set_price(-100)

print("\nUpdated Product Info:")
product.get_info()
```

## Task 3: Inheritance

```python
# Task 3: Inheritance (Single-Level)
# ElectronicProduct inherits from Product base class

class ElectronicProduct(Product):
    """ElectronicProduct class inheriting from Product"""
    
    def __init__(self, name, price, category, warranty_years):
        """Initialize electronic product with warranty_years attribute"""
        super().__init__(name, price, category)
        self.warranty_years = warranty_years
    
    def get_info(self):
        """Override get_info to include warranty info"""
        super().get_info()
        print(f"Warranty: {self.warranty_years} years")

print("Task 3: Inheritance (Single-Level)")
print("=" * 50)

# Create ElectronicProduct object
laptop = ElectronicProduct("Dell XPS", 1299, "Electronics", 2)

print("\nElectronicProduct Object (inherits from Product):")
laptop.get_info()

print("\n" + "=" * 50)
print("Demonstrating Inheritance + Method Overriding:")
print("=" * 50)

# Access inherited attributes and methods
print(f"\nProduct Name: {laptop.name}")
print(f"Product Price: ${laptop.price}")
print(f"Product Category: {laptop.category}")
print(f"Additional Attribute - Warranty: {laptop.warranty_years} years")

# Call overridden method
print("\nCalling overridden get_info() method:")
laptop.get_info()
```

## Task 5: Abstraction (Using Abstract Base Class)

```python
from abc import ABC, abstractmethod

# Task 5: Abstraction using Abstract Base Class

class Payment(ABC):
    """Abstract Payment class with abstract method"""
    
    @abstractmethod
    def process_payment(self, amount):
        """Abstract method for processing payment"""
        pass

class CreditCardPayment(Payment):
    """CreditCard payment implementation"""
    
    def __init__(self, card_number):
        """Initialize with card number"""
        self.card_number = card_number
    
    def process_payment(self, amount):
        """Override: Process payment via credit card"""
        print(f"💳 Processing Credit Card Payment")
        print(f"   Card Number: {self.card_number[-4:].rjust(16, '*')}")
        print(f"   Amount: ${amount:.2f}")
        print(f"   Status: ✓ Payment Successful")

class UPIPayment(Payment):
    """UPI payment implementation"""
    
    def __init__(self, upi_id):
        """Initialize with UPI ID"""
        self.upi_id = upi_id
    
    def process_payment(self, amount):
        """Override: Process payment via UPI"""
        print(f"📲 Processing UPI Payment")
        print(f"   UPI ID: {self.upi_id}")
        print(f"   Amount: ${amount:.2f}")
        print(f"   Status: ✓ Payment Successful")

print("Task 5: Abstraction (Using Abstract Base Class)")
print("=" * 50)

# Create instances of concrete payment classes
credit_payment = CreditCardPayment("1234567890123456")
upi_payment = UPIPayment("user@paytm")

# Test all payment methods
print("\nTesting CreditCardPayment:")
print("-" * 50)
credit_payment.process_payment(150.00)

print("\n\nTesting UPIPayment:")
print("-" * 50)
upi_payment.process_payment(150.00)

print("\n\n" + "=" * 50)
print("Note: Both classes implement the abstract method")
print("process_payment() from the Payment abstract class.")
```

## Task 4: Polymorphism (Method Overriding)

```python
# Task 4: Polymorphism
# Create Laptop and Mobile classes that inherit from Product

class Laptop(Product):
    """Laptop class with custom get_info implementation"""
    
    def __init__(self, name, price, category, processor):
        """Initialize Laptop with processor attribute"""
        super().__init__(name, price, category)
        self.processor = processor
    
    def get_info(self):
        """Override get_info with Laptop-specific style"""
        print(f"🖥️  LAPTOP DETAILS 🖥️")
        print(f"   Name: {self.name}")
        print(f"   Price: ${self.price}")
        print(f"   Category: {self.category}")
        print(f"   Processor: {self.processor}")

class Mobile(Product):
    """Mobile class with custom get_info implementation"""
    
    def __init__(self, name, price, category, storage):
        """Initialize Mobile with storage attribute"""
        super().__init__(name, price, category)
        self.storage = storage
    
    def get_info(self):
        """Override get_info with Mobile-specific style"""
        print(f"📱 MOBILE DETAILS 📱")
        print(f"   Name: {self.name}")
        print(f"   Price: ${self.price}")
        print(f"   Category: {self.category}")
        print(f"   Storage: {self.storage}GB")

print("Task 4: Polymorphism")
print("=" * 50)

# Create objects of Laptop and Mobile
laptop = Laptop("MacBook Pro", 1999, "Electronics", "M2 Max")
mobile = Mobile("iPhone 15", 999, "Electronics", 256)

# Demonstrate polymorphism with a loop
print("\nIterating over objects and calling get_info():")
print("=" * 50)

devices = [laptop, mobile]

for device in devices:
    print()
    device.get_info()
    print()

print("=" * 50)
print("Note: Same method name (get_info) behaves differently")
print("for different object types - This is POLYMORPHISM!")
```

## Task 6: Magic Methods & Operator Overloading

```python
# Task 6: Magic Methods & Operator Overloading
# Enhanced Product class with __str__ and __add__ magic methods

class ProductMagic:
    """Product class with magic methods"""
    
    def __init__(self, name, price, category):
        """Initialize product attributes"""
        self.name = name
        self.price = price
        self.category = category
    
    def __str__(self):
        """Magic method __str__: Returns readable string representation"""
        return f"Product({self.name}, ${self.price}, {self.category})"
    
    def __repr__(self):
        """Magic method __repr__: Returns detailed representation"""
        return f"ProductMagic('{self.name}', {self.price}, '{self.category}')"
    
    def __add__(self, other):
        """Magic method __add__: Operator overloading for + 
        Returns the total combined price of two products"""
        if isinstance(other, ProductMagic):
            return self.price + other.price
        elif isinstance(other, (int, float)):
            return self.price + other
        return NotImplemented

print("Task 6: Magic Methods & Operator Overloading")
print("=" * 50)

# Create product objects
product1 = ProductMagic("Laptop", 1200, "Electronics")
product2 = ProductMagic("Mouse", 50, "Electronics")
product3 = ProductMagic("Keyboard", 80, "Electronics")

print("\nProduct 1 - Using __str__ method:")
print(f"  {product1}")

print("\nProduct 2 - Using __str__ method:")
print(f"  {product2}")

print("\n" + "=" * 50)
print("Operator Overloading (__add__) Test:")
print("=" * 50)

# Test operator overloading - pairwise additions
print(f"\nProduct 1 Price: ${product1.price}")
print(f"Product 2 Price: ${product2.price}")
print(f"Product 3 Price: ${product3.price}")

combined_price_1_2 = product1 + product2
print(f"\nUsing product1 + product2:")
print(f"  Combined Price: ${combined_price_1_2}")

combined_price_2_3 = product2 + product3
print(f"\nUsing product2 + product3:")
print(f"  Combined Price: ${combined_price_2_3}")

print("\n" + "=" * 50)
print("Testing with Multiple Products:")
print("=" * 50)

# Calculate total of all three products by adding successive results
total_price = product1 + product2
total_price = total_price + product3.price

print(f"\nProduct1 + Product2 + Product3 = ${total_price}")
```

## Task 7: Mini Project - Simple Inventory System (OOP Only)

```python
# Task 7: Mini Project - Simple Inventory System (Using OOP Only)
# Create Inventory and Store classes

class Product:
    """Base Product class for inventory system"""
    
    def __init__(self, name, price):
        """Initialize product with name and price"""
        self.name = name
        self.price = price
    
    def __str__(self):
        """String representation of product"""
        return f"{self.name} (${self.price})"
    
    def __add__(self, other):
        """Add two product prices or add to a number"""
        if isinstance(other, Product):
            return self.price + other.price
        elif isinstance(other, (int, float)):
            return self.price + other
        return NotImplemented
    
    def __radd__(self, other):
        """Support right addition (int + Product)"""
        if isinstance(other, (int, float)):
            return other + self.price
        return NotImplemented

class Inventory:
    """Inventory class for managing products"""
    
    def __init__(self):
        """Initialize empty products list"""
        self.products = []
    
    def add_product(self, product):
        """Add a product to inventory"""
        self.products.append(product)
        print(f"✓ Added: {product.name} - ${product.price}")
    
    def remove_product(self, name):
        """Remove a product from inventory by name"""
        for product in self.products:
            if product.name.lower() == name.lower():
                self.products.remove(product)
                print(f"✓ Removed: {name}")
                return True
        print(f"✗ Product '{name}' not found")
        return False
    
    def get_total_value(self):
        """Calculate total value of all products in inventory"""
        total = 0
        for product in self.products:
            total += product.price
        return total
    
    def show_all_products(self):
        """Display all products in inventory"""
        if not self.products:
            print("Inventory is empty")
            return
        
        print("\n" + "=" * 50)
        print("📦 INVENTORY CONTENTS 📦")
        print("=" * 50)
        for i, product in enumerate(self.products, 1):
            print(f"{i}. {product}")
        print("=" * 50)

class Store:
    """Store class for managing inventory and sales"""
    
    def __init__(self, store_name):
        """Initialize store with name and inventory"""
        self.store_name = store_name
        self.inventory = Inventory()
    
    def add_new_product(self, name, price):
        """Create and add a new product to store"""
        product = Product(name, price)
        self.inventory.add_product(product)
        return product
    
    def show_summary(self):
        """Display store summary with inventory details"""
        print("\n" + "=" * 50)
        print(f"🏪 {self.store_name.upper()} SUMMARY 🏪")
        print("=" * 50)
        
        total_items = len(self.inventory.products)
        total_value = self.inventory.get_total_value()
        
        print(f"Total Products: {total_items}")
        print(f"Total Inventory Value: ${total_value:.2f}")
        
        if total_items > 0:
            avg_price = total_value / total_items
            print(f"Average Price per Product: ${avg_price:.2f}")
        print("=" * 50)

# Test the Inventory System
print("Task 7: Mini Project - Simple Inventory System")
print("=" * 50)

# 1. Create a Store object
store = Store("Tech Store")

# 2. Add 3 products
print("\n1️⃣ Adding 3 Products to Store:")
print("=" * 50)
product1 = store.add_new_product("Laptop", 1200)
product2 = store.add_new_product("Smartphone", 800)
product3 = store.add_new_product("Tablet", 500)

# 3. Display all products
print("\n2️⃣ Display All Products:")
store.inventory.show_all_products()

# 4. Show store summary
print("\n3️⃣ Store Summary:")
store.show_summary()

# 5. Using __add__ to combine prices of two products
print("\n4️⃣ Combining Product Prices Using __add__:")
print("=" * 50)
combined_1_2 = product1 + product2
combined_2_3 = product2 + product3

# Calculate total using step-by-step additions
total_step1 = product1 + product2  # Returns int: 2000
total_all = total_step1 + product3.price  # Add remaining product price

print(f"Laptop + Smartphone: ${combined_1_2}")
print(f"Smartphone + Tablet: ${combined_2_3}")
print(f"Laptop + Smartphone + Tablet: ${total_all}")
print("=" * 50)
```
