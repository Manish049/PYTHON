## Task 1: Safe Division Utility

```python
# Task 1: Safe Division Utility
# Program that takes two inputs and handles ValueError and ZeroDivisionError

try:
    # Get inputs from user
    numerator = float(input("Enter numerator: "))
    denominator = float(input("Enter denominator: "))
    
    # Perform division
    result = numerator / denominator
    
except ValueError:
    print("Error: Invalid input. Please enter numeric values only.")
    
except ZeroDivisionError:
    print("Error: Cannot divide by zero. Denominator must be non-zero.")
    
else:
    # Print result only if no error occurs
    print(f"Result: {numerator} / {denominator} = {result}")
    
finally:
    # Always execute this block
    print("Operation Complete")
```

### Test Cases

Test 1: Valid division (10 / 2)

```python
# Function-based version for testing different scenarios
def safe_division(numerator, denominator):
    """Perform safe division with exception handling"""
    try:
        numerator = float(numerator)
        denominator = float(denominator)
        
        result = numerator / denominator
        
    except ValueError:
        print(f"Error: Invalid input. Please enter numeric values only.")
        
    except ZeroDivisionError:
        print(f"Error: Cannot divide by zero. Denominator must be non-zero.")
        
    else:
        print(f"Result: {numerator} / {denominator} = {result}")
        
    finally:
        print("Operation Complete\n")

# Test Case 1: Valid division
print("Test 1: Valid Division (10 / 2)")
safe_division(10, 2)

# Test Case 2: Division by zero
print("Test 2: Division by Zero (10 / 0)")
safe_division(10, 0)

# Test Case 3: Invalid input (non-numeric)
print("Test 3: Invalid Input ('abc' / 2)")
safe_division("abc", 2)

# Test Case 4: Decimal division
print("Test 4: Decimal Division (7.5 / 2.5)")
safe_division(7.5, 2.5)
```

## Task 2: Bill Calculator with Error Handling

Given a list of product prices: `prices = [120, 350, 'abc', 500, -200, 800]`

Write code that:
1. Iterates through the list
2. Tries to add only valid (positive numeric) prices to the total
3. Handles:
   - TypeError if value is not a number
   - Custom exception using raise ValueError("Negative price not allowed")
4. Prints the final total
5. Expected behavior: Skip invalid items but continue processing

```python
# Task 2: Bill Calculator with Error Handling
prices = [120, 350, 'abc', 500, -200, 800]

total = 0

for price in prices:
    try:
        # Try to convert to float
        price_value = float(price)
        
        # Check if price is negative
        if price_value < 0:
            raise ValueError("Negative price not allowed")
        
        # Add valid price to total
        total += price_value
        print(f"Added: {price_value}")
        
    except ValueError as e:
        if "Negative price not allowed" in str(e):
            print(f"Error: {e} - Skipping price: {price}")
        else:
            print(f"Error: Invalid price format '{price}' - Skipping")
            
    except TypeError:
        print(f"Error: Type error for price '{price}' - Skipping")

print(f"\nFinal Total: {total}")
```

## Task 3: Custom Exception - Age Validator

Write a function `check_age(age)` that:
1. Raises a custom exception ValueError("Age must be between 1 and 120") if age is out of range
2. In main code:
   - Take age input from user
   - Use try-except to catch and print the custom error message

```python
# Task 3: Custom Exception - Age Validator
def check_age(age):
    """Validate age is between 1 and 120"""
    try:
        age_value = int(age)
        
        if age_value < 1 or age_value > 120:
            raise ValueError("Age must be between 1 and 120")
        
        return age_value
        
    except ValueError as e:
        if "Age must be between" in str(e):
            raise
        else:
            raise ValueError("Age must be a valid number")

# Test cases
print("Age Validator Tests:\n")

test_ages = [25, 150, -5, "abc", 120, 1, 0]

for age in test_ages:
    try:
        result = check_age(age)
        print(f"Age {age}: Valid ✓")
    except ValueError as e:
        print(f"Age {age}: Error - {e}")
```

## Task 4: File Reader with Exception Handling

Write a program that:
1. Asks the user for a filename
2. Tries to open and read the file
3. Handles:
   - FileNotFoundError
   - PermissionError
4. If successful, print first 3 lines of the file
5. Use finally to print "File operation attempted"

```python
# Task 4: File Reader with Exception Handling

def read_file(filename):
    """Try to open and read a file"""
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()
            
            # Print first 3 lines (or fewer if file is shorter)
            print(f"First {min(3, len(lines))} lines of '{filename}':\n")
            for i, line in enumerate(lines[:3]):
                print(f"{i+1}: {line.rstrip()}")
                
    except FileNotFoundError:
        print(f"Error: File '{filename}' not found.")
        
    except PermissionError:
        print(f"Error: Permission denied. Cannot read file '{filename}'.")
        
    except Exception as e:
        print(f"Error: An unexpected error occurred - {e}")
        
    finally:
        print("File operation attempted")

# Demo with test files
print("Test 1: Reading an existing file")
print("=" * 40)
read_file("Math_utils.py")

print("\n\nTest 2: Reading a non-existent file")
print("=" * 40)
read_file("nonexistent_file.txt")

print("\n\nTest 3: Interactive version (commented)")
print("=" * 40)
print("""
# To use the interactive version, uncomment:
# filename = input("Enter filename: ")
# read_file(filename)
""")
```
