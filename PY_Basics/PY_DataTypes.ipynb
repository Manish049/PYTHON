# Python Data Types - Complete Training Guide

## 📚 Overview
Python supports several built-in data types to store different kinds of values. This notebook covers all major data types with practical examples and hands-on exercises.

### Data Types Categories:
1. **Numeric Types**: int, float, complex
2. **Sequence Types**: str, list, tuple, range
3. **Mapping Type**: dict
4. **Set Types**: set, frozenset
5. **Boolean Type**: bool
6. **Special Type**: NoneType

## Section 1: Numeric Data Types

### 1.1 Integer (int)
- Whole numbers without decimal points
- Can be positive, negative, or zero
- Size limited only by available memory

```python
# Section 1.1: Integer Data Type Examples

# Creating integers
age = 25
price = 1200
negative_num = -100
large_num = 999999999999

print("Integer Examples:")
print(f"age = {age}, type: {type(age)}")
print(f"price = {price}, type: {type(price)}")
print(f"negative_num = {negative_num}, type: {type(negative_num)}")
print(f"large_num = {large_num}, type: {type(large_num)}")

# Integer operations
print("\n" + "="*50)
print("Integer Operations:")
print("="*50)

a = 10
b = 3

print(f"\nAddition: {a} + {b} = {a + b}")
print(f"Subtraction: {a} - {b} = {a - b}")
print(f"Multiplication: {a} * {b} = {a * b}")
print(f"Division: {a} / {b} = {a / b}")
print(f"Floor Division: {a} // {b} = {a // b}")
print(f"Modulus: {a} % {b} = {a % b}")
print(f"Exponentiation: {a} ** {b} = {a ** b}")

# Integer in different bases
print("\n" + "="*50)
print("Integers in Different Bases:")
print("="*50)

binary = 0b1010  # Binary (base 2)
octal = 0o12     # Octal (base 8)
hexadecimal = 0xA  # Hexadecimal (base 16)

print(f"Binary 0b1010 = {binary}")
print(f"Octal 0o12 = {octal}")
print(f"Hexadecimal 0xA = {hexadecimal}")
```

### 1.2 Float (float)
- Numbers with decimal points
- Can represent fractional values
- Used for precise calculations

```python
# Section 1.2: Float Data Type Examples

# Creating floats
height = 5.9
weight = 75.5
pi = 3.14159
scientific = 1.23e-4  # Scientific notation

print("Float Examples:")
print(f"height = {height}, type: {type(height)}")
print(f"weight = {weight}, type: {type(weight)}")
print(f"pi = {pi}, type: {type(pi)}")
print(f"scientific notation 1.23e-4 = {scientific}")

# Float operations
print("\n" + "="*50)
print("Float Operations:")
print("="*50)

x = 10.5
y = 3.2

print(f"\nAddition: {x} + {y} = {x + y}")
print(f"Subtraction: {x} - {y} = {x - y}")
print(f"Multiplication: {x} * {y} = {x * y}")
print(f"Division: {x} / {y} = {x / y}")
print(f"Floor Division: {x} // {y} = {x // y}")
print(f"Modulus: {x} % {y} = {x % y}")

# Float precision
print("\n" + "="*50)
print("Float Precision & Rounding:")
print("="*50)

num = 3.14159265359
print(f"Original: {num}")
print(f"Round to 2 decimals: {round(num, 2)}")
print(f"Round to 4 decimals: {round(num, 4)}")
print(f"Floor: {int(num)}")

# Special float values
print("\n" + "="*50)
print("Special Float Values:")
print("="*50)

print(f"Positive Infinity: {float('inf')}")
print(f"Negative Infinity: {float('-inf')}")
print(f"Not a Number (NaN): {float('nan')}")
```

### 1.3 Complex (complex)
- Numbers with real and imaginary parts
- Format: real + imaginary*j
- Used in scientific and engineering calculations

```python
# Section 1.3: Complex Number Examples

# Creating complex numbers
c1 = 3 + 4j
c2 = 5 - 2j
c3 = complex(2, 3)  # Using complex() constructor

print("Complex Number Examples:")
print(f"c1 = {c1}, type: {type(c1)}")
print(f"c2 = {c2}, type: {type(c2)}")
print(f"c3 = complex(2, 3) = {c3}")

# Complex number properties
print("\n" + "="*50)
print("Complex Number Properties:")
print("="*50)

print(f"\nReal part of {c1}: {c1.real}")
print(f"Imaginary part of {c1}: {c1.imag}")
print(f"Conjugate of {c1}: {c1.conjugate()}")

# Complex arithmetic
print("\n" + "="*50)
print("Complex Number Operations:")
print("="*50)

c1 = 3 + 4j
c2 = 1 + 2j

print(f"\nc1 = {c1}")
print(f"c2 = {c2}")
print(f"\nAddition: {c1} + {c2} = {c1 + c2}")
print(f"Subtraction: {c1} - {c2} = {c1 - c2}")
print(f"Multiplication: {c1} * {c2} = {c1 * c2}")
print(f"Division: {c1} / {c2} = {c1 / c2}")
print(f"Power: {c1} ** 2 = {c1 ** 2}")

# Magnitude of complex number
print("\n" + "="*50)
print("Complex Number Magnitude:")
print("="*50)

print(f"\nMagnitude of {c1}: {abs(c1)}")
print(f"Magnitude of {c2}: {abs(c2)}")
```

## Section 2: String Data Type (str)

### 2.1 String Basics
- Sequence of characters enclosed in quotes
- Can use single quotes, double quotes, or triple quotes
- Immutable - cannot be changed after creation
- Indexed - each character has a position

```python
# Section 2: String Data Type Examples

# Creating strings
single_quote = 'Hello'
double_quote = "World"
triple_quote = """This is a
multi-line string that can span
multiple lines"""

print("String Creation Examples:")
print(f"single_quote: {single_quote}, type: {type(single_quote)}")
print(f"double_quote: {double_quote}")
print(f"\ntriple_quote:\n{triple_quote}")

# String indexing and slicing
print("\n" + "="*50)
print("String Indexing & Slicing:")
print("="*50)

text = "Python"
print(f"\nString: {text}")
print(f"Length: {len(text)}")
print(f"First character (index 0): {text[0]}")
print(f"Last character (index -1): {text[-1]}")
print(f"Characters 0-2 (slice [0:3]): {text[0:3]}")
print(f"Characters from index 2 onwards: {text[2:]}")
print(f"Characters up to index 2: {text[:2]}")
print(f"Every 2nd character: {text[::2]}")
print(f"Reversed string: {text[::-1]}")

# String concatenation
print("\n" + "="*50)
print("String Concatenation & Repetition:")
print("="*50)

first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name

print(f"\nfirst_name = '{first_name}'")
print(f"last_name = '{last_name}'")
print(f"Concatenation: '{full_name}'")
print(f"Repetition: '{'-' * 20}'")

# String methods
print("\n" + "="*50)
print("Common String Methods:")
print("="*50)

text = "Hello World"
print(f"\nOriginal: '{text}'")
print(f"Upper case: '{text.upper()}'")
print(f"Lower case: '{text.lower()}'")
print(f"Title case: '{text.title()}'")
print(f"Replace 'World' with 'Python': '{text.replace('World', 'Python')}'")
print(f"Split by space: {text.split(' ')}")
print(f"Starts with 'Hello': {text.startswith('Hello')}")
print(f"Ends with 'World': {text.endswith('World')}")
print(f"Find index of 'World': {text.find('World')}")

# String formatting
print("\n" + "="*50)
print("String Formatting:")
print("="*50)

name = "Alice"
age = 25
score = 95.5

# Method 1: f-string (modern)
print(f"\nF-string: Hello {name}, you are {age} years old and your score is {score}")

# Method 2: .format()
print("Format method: Hello {}, you are {} years old and your score is {}".format(name, age, score))

# Method 3: % formatting
print("Percent formatting: Hello %s, you are %d years old and your score is %.1f" % (name, age, score))

# Method 4: Variable formatting  
print(f"F-string with format spec: Score: {score:.2f}")
```

## Section 3: Boolean Data Type (bool)

### 3.1 Boolean Basics
- Only two values: True and False
- Result of comparison and logical operations
- Used in conditional statements
- Important for control flow

```python
# Section 3: Boolean Data Type Examples

# Boolean values
is_student = True
is_employed = False

print("Boolean Examples:")
print(f"is_student = {is_student}, type: {type(is_student)}")
print(f"is_employed = {is_employed}, type: {type(is_employed)}")

# Comparison operators
print("\n" + "="*50)
print("Comparison Operators:")
print("="*50)

a = 10
b = 20

print(f"\na = {a}, b = {b}")
print(f"a == b: {a == b}")
print(f"a != b: {a != b}")
print(f"a < b: {a < b}")
print(f"a > b: {a > b}")
print(f"a <= b: {a <= b}")
print(f"a >= b: {a >= b}")

# Logical operators
print("\n" + "="*50)
print("Logical Operators:")
print("="*50)

x = True
y = False

print(f"\nx = {x}, y = {y}")
print(f"x and y: {x and y}")
print(f"x or y: {x or y}")
print(f"not x: {not x}")
print(f"not y: {not y}")

# Combining boolean conditions
print("\n" + "="*50)
print("Complex Boolean Expressions:")
print("="*50)

age = 25
income = 50000
has_job = True

print(f"\nage = {age}, income = {income}, has_job = {has_job}")

can_apply_credit = (age >= 18) and (income >= 30000) and has_job
print(f"Can apply for credit card: {can_apply_credit}")

is_eligible = (age < 18) or (income < 30000) or not has_job
print(f"Is eligible for student discount: {is_eligible}")

# Boolean conversions
print("\n" + "="*50)
print("Boolean Conversions (Truthy/Falsy):")
print("="*50)

print(f"\nbool(1): {bool(1)}")
print(f"bool(0): {bool(0)}")
print(f"bool('Hello'): {bool('Hello')}")
print(f"bool(''): {bool('')}")
print(f"bool([1, 2, 3]): {bool([1, 2, 3])}")
print(f"bool([]): {bool([])}")
print(f"bool(None): {bool(None)}")
```

## Section 4: List Data Type

### 4.1 List Basics
- Ordered collection of items
- Items can be of different types
- Mutable - can be modified
- Indexed - access items by position
- Enclosed in square brackets []

```python
# Section 4: List Data Type Examples

# Creating lists
numbers = [1, 2, 3, 4, 5]
mixed_list = [1, "hello", 3.14, True, None]
empty_list = []

print("List Creation Examples:")
print(f"numbers = {numbers}, type: {type(numbers)}")
print(f"mixed_list = {mixed_list}")
print(f"empty_list = {empty_list}")

# List indexing and slicing
print("\n" + "="*50)
print("List Indexing & Slicing:")
print("="*50)

fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print(f"\nList: {fruits}")
print(f"Length: {len(fruits)}")
print(f"First item: {fruits[0]}")
print(f"Last item: {fruits[-1]}")
print(f"Items 1-3 (slice [1:3]): {fruits[1:3]}")
print(f"Items from index 2: {fruits[2:]}")
print(f"Items up to index 2: {fruits[:2]}")
print(f"Every 2nd item: {fruits[::2]}")
print(f"Reversed list: {fruits[::-1]}")

# Modifying lists
print("\n" + "="*50)
print("Modifying Lists:")
print("="*50)

colors = ["red", "green", "blue"]
print(f"\nOriginal list: {colors}")

# Change an item
colors[0] = "yellow"
print(f"After changing index 0 to 'yellow': {colors}")

# Add items
colors.append("purple")
print(f"After append('purple'): {colors}")

colors.insert(1, "orange")
print(f"After insert(1, 'orange'): {colors}")

# Remove items
colors.remove("green")
print(f"After remove('green'): {colors}")

popped = colors.pop()
print(f"After pop(): {colors}, popped item: {popped}")

# List operations
print("\n" + "="*50)
print("List Operations:")
print("="*50)

list1 = [1, 2, 3]
list2 = [4, 5, 6]

print(f"\nlist1 = {list1}")
print(f"list2 = {list2}")
print(f"Concatenation (list1 + list2): {list1 + list2}")
print(f"Repetition (list1 * 3): {list1 * 3}")
print(f"Is 2 in list1? {2 in list1}")

# List methods
print("\n" + "="*50)
print("Common List Methods:")
print("="*50)

nums = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"\nOriginal: {nums}")

print(f"count(1): {nums.count(1)}")
print(f"index(4): {nums.index(4)}")

nums_copy = nums.copy()
nums_copy.sort()
print(f"sort(): {nums_copy}")

nums_copy.reverse()
print(f"reverse(): {nums_copy}")

nums.clear()
print(f"After clear(): {nums}")
```

## Section 5: Tuple Data Type

### 5.1 Tuple Basics
- Ordered collection of items similar to lists
- Immutable - cannot be modified after creation
- Indexed - access items by position
- Can contain items of different types
- Enclosed in parentheses () or just comma-separated
- More memory efficient than lists

```python
# Section 5: Tuple Data Type Examples

# Creating tuples
numbers = (1, 2, 3, 4, 5)
mixed_tuple = (1, "hello", 3.14, True)
single_item = (42,)  # Note the comma for single item
empty_tuple = ()
tuple_from_list = tuple([1, 2, 3])

print("Tuple Creation Examples:")
print(f"numbers = {numbers}, type: {type(numbers)}")
print(f"mixed_tuple = {mixed_tuple}")
print(f"single_item = {single_item}")
print(f"empty_tuple = {empty_tuple}")
print(f"tuple_from_list = {tuple_from_list}")

# Tuple indexing and slicing
print("\n" + "="*50)
print("Tuple Indexing & Slicing:")
print("="*50)

colors = ("red", "green", "blue", "yellow", "purple")
print(f"\nTuple: {colors}")
print(f"Length: {len(colors)}")
print(f"First item: {colors[0]}")
print(f"Last item: {colors[-1]}")
print(f"Items 1-3 (slice [1:3]): {colors[1:3]}")
print(f"Items from index 2: {colors[2:]}")
print(f"Every 2nd item: {colors[::2]}")
print(f"Reversed tuple: {colors[::-1]}")

# Attempting to modify (immutable)
print("\n" + "="*50)
print("Tuple Immutability:")
print("="*50)

my_tuple = (10, 20, 30)
print(f"\nTuple: {my_tuple}")
print("Attempting to modify...")
try:
    my_tuple[0] = 100
except TypeError as e:
    print(f"Error: {e}")

# Tuple operations
print("\n" + "="*50)
print("Tuple Operations:")
print("="*50)

t1 = (1, 2, 3)
t2 = (4, 5)

print(f"\nt1 = {t1}")
print(f"t2 = {t2}")
print(f"Concatenation (t1 + t2): {t1 + t2}")
print(f"Repetition (t1 * 2): {t1 * 2}")
print(f"Is 2 in t1? {2 in t1}")

# Tuple unpacking
print("\n" + "="*50)
print("Tuple Unpacking:")
print("="*50)

coordinates = (10, 20, 30)
x, y, z = coordinates
print(f"\ncoordinates = {coordinates}")
print(f"After unpacking: x={x}, y={y}, z={z}")

# Nested unpacking
person = ("John", 25, ("New York", "USA"))
name, age, (city, country) = person
print(f"\nperson = {person}")
print(f"After unpacking: name={name}, age={age}, city={city}, country={country}")

# Tuple methods
print("\n" + "="*50)
print("Tuple Methods:")
print("="*50)

nums = (1, 2, 3, 2, 4, 2, 5)
print(f"\nTuple: {nums}")
print(f"count(2): {nums.count(2)}")
print(f"index(3): {nums.index(3)}")

# Convert tuple to list and back
print("\n" + "="*50)
print("Convert Between List and Tuple:")
print("="*50)

my_list = [1, 2, 3, 4]
my_tuple = tuple(my_list)
back_to_list = list(my_tuple)

print(f"\nOriginal list: {my_list}")
print(f"Converted to tuple: {my_tuple}")
print(f"Converted back to list: {back_to_list}")
```

## Section 6: Dictionary Data Type

### 6.1 Dictionary Basics
- Unordered collection of key-value pairs
- Mutable - can be modified
- Indexed by keys, not by position
- Keys must be unique and immutable
- Enclosed in curly braces {}
- Efficient lookup by key

```python
# Section 6: Dictionary Data Type Examples

# Creating dictionaries
student = {"name": "John", "age": 20, "major": "Computer Science"}
empty_dict = {}
dict_constructor = dict(name="Alice", age=22)

print("Dictionary Creation Examples:")
print(f"student = {student}, type: {type(student)}")
print(f"empty_dict = {empty_dict}")
print(f"dict_constructor = {dict_constructor}")

# Accessing dictionary items
print("\n" + "="*50)
print("Accessing Dictionary Items:")
print("="*50)

person = {"name": "John", "age": 25, "city": "New York", "email": "john@example.com"}
print(f"\nDictionary: {person}")
print(f"person['name']: {person['name']}")
print(f"person['age']: {person['age']}")
print(f"person.get('city'): {person.get('city')}")
print(f"person.get('country', 'USA'): {person.get('country', 'USA')}")  # Default value

# Modifying dictionaries
print("\n" + "="*50)
print("Modifying Dictionaries:")
print("="*50)

book = {"title": "Python 101", "author": "John Doe", "year": 2020}
print(f"\nOriginal: {book}")

# Change value
book['year'] = 2021
print(f"After changing year: {book}")

# Add new item
book['pages'] = 300
print(f"After adding pages: {book}")

# Update multiple items
book.update({"genre": "Programming", "rating": 4.5})
print(f"After update(): {book}")

# Remove items
deleted = book.pop('pages')
print(f"After pop('pages'): {book}, deleted value: {deleted}")

# Dictionary operations
print("\n" + "="*50)
print("Dictionary Operations:")
print("="*50)

car = {"brand": "Toyota", "model": "Camry", "year": 2022, "color": "Blue"}
print(f"\nDictionary: {car}")
print(f"Keys: {list(car.keys())}")
print(f"Values: {list(car.values())}")
print(f"Items: {list(car.items())}")
print(f"Number of items: {len(car)}")
print(f"Is 'brand' in car? {'brand' in car}")
print(f"Is 'price' in car? {'price' in car}")

# Iterating through dictionaries
print("\n" + "="*50)
print("Iterating Through Dictionaries:")
print("="*50)

scores = {"Alice": 95, "Bob": 87, "Charlie": 92}
print(f"\nDictionary: {scores}")

print("\nIterating through keys:")
for key in scores:
    print(f"  {key}: {scores[key]}")

print("\nIterating through items:")
for key, value in scores.items():
    print(f"  {key}: {value}")

# Nested dictionaries
print("\n" + "="*50)
print("Nested Dictionaries:")
print("="*50)

company = {
    "name": "TechCorp",
    "employees": 100,
    "departments": {
        "IT": {"count": 25, "manager": "John"},
        "HR": {"count": 10, "manager": "Jane"},
        "Sales": {"count": 30, "manager": "Bob"}
    }
}

print(f"\nCompany: {company['name']}")
print(f"IT Department: {company['departments']['IT']}")
print(f"IT Manager: {company['departments']['IT']['manager']}")
print(f"HR Count: {company['departments']['HR']['count']}")

# Dictionary methods
print("\n" + "="*50)
print("Common Dictionary Methods:")
print("="*50)

original = {"a": 1, "b": 2, "c": 3}
copy = original.copy()
print(f"copy(): {copy}")

original.clear()
print(f"After clear(): {original}")

d1 = {"x": 1, "y": 2}
d2 = {"y": 3, "z": 4}
d1.update(d2)
print(f"After update({d2}): {d1}")
```

## Section 7: Set Data Type

### 7.1 Set Basics
- Unordered collection of unique items
- Mutable - can be modified
- Items cannot be duplicated
- No indexing - access by membership test
- Useful for removing duplicates and set operations
- Enclosed in curly braces {} or set() constructor
- frozenset is an immutable version

```python
# Section 7: Set Data Type Examples

# Creating sets
numbers = {1, 2, 3, 4, 5}
mixed_set = {1, "hello", 3.14, True}
empty_set = set()  # Note: {} creates empty dict, not set
set_from_list = set([1, 2, 2, 3, 3, 4])  # Duplicates removed

print("Set Creation Examples:")
print(f"numbers = {numbers}, type: {type(numbers)}")
print(f"mixed_set = {mixed_set}")
print(f"empty_set = {empty_set}")
print(f"set_from_list = {set_from_list} (duplicates removed)")

# Set characteristics
print("\n" + "="*50)
print("Set Characteristics:")
print("="*50)

original_list = [1, 2, 2, 3, 3, 3, 4, 5, 5]
unique_set = set(original_list)
print(f"\nOriginal list: {original_list}")
print(f"Converted to set: {unique_set}")
print(f"Length: {len(unique_set)}")

# Add and remove items
print("\n" + "="*50)
print("Modifying Sets:")
print("="*50)

colors = {"red", "green", "blue"}
print(f"\nOriginal: {colors}")

colors.add("yellow")
print(f"After add('yellow'): {colors}")

colors.update(["purple", "orange"])
print(f"After update(['purple', 'orange']): {colors}")

colors.remove("green")
print(f"After remove('green'): {colors}")

removed = colors.pop()
print(f"After pop(): {colors}, removed item: {removed}")

# Set operations
print("\n" + "="*50)
print("Set Mathematical Operations:")
print("="*50)

set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}

print(f"\nset1 = {set1}")
print(f"set2 = {set2}")

# Union
print(f"\nUnion (set1 | set2): {set1 | set2}")
print(f"Union (set1.union(set2)): {set1.union(set2)}")

# Intersection
print(f"Intersection (set1 & set2): {set1 & set2}")
print(f"Intersection (set1.intersection(set2)): {set1.intersection(set2)}")

# Difference
print(f"Difference (set1 - set2): {set1 - set2}")
print(f"Difference (set1.difference(set2)): {set1.difference(set2)}")

# Symmetric difference
print(f"Symmetric Difference (set1 ^ set2): {set1 ^ set2}")

# Subset and Superset
print("\n" + "="*50)
print("Subset & Superset:")
print("="*50)

a = {1, 2}
b = {1, 2, 3, 4}

print(f"\na = {a}")
print(f"b = {b}")
print(f"Is a subset of b? {a.issubset(b)}")
print(f"Is b superset of a? {b.issuperset(a)}")
print(f"Are they disjoint? {a.isdisjoint(b)}")

# Set membership
print("\n" + "="*50)
print("Set Membership Testing:")
print("="*50)

fruits = {"apple", "banana", "cherry"}
print(f"\nSet: {fruits}")
print(f"'apple' in fruits: {'apple' in fruits}")
print(f"'orange' in fruits: {'orange' in fruits}")

# Frozenset (immutable)
print("\n" + "="*50)
print("Frozenset (Immutable Sets):")
print("="*50)

my_set = frozenset([1, 2, 3, 4, 5])
print(f"\nFrozenset: {my_set}")
print(f"Type: {type(my_set)}")

print("Attempting to modify frozenset...")
try:
    my_set.add(6)
except AttributeError as e:
    print(f"Error: Cannot modify frozenset (immutable)")

# Can be used in sets and as dict keys
print(f"Can use as dict key: dict({{my_set: 'value'}})")
```

## Section 8: Range Data Type

### 8.1 Range Basics
- Represents a sequence of numbers
- Immutable - cannot be modified
- Memory efficient - doesn't store all values
- Used primarily in loops
- Syntax: range(start, stop, step)

```python
# Section 8: Range Data Type Examples

# Creating ranges
range1 = range(5)           # 0 to 4
range2 = range(2, 8)        # 2 to 7
range3 = range(0, 10, 2)    # 0 to 9, step 2
range4 = range(10, 0, -1)   # 10 to 1, step -1

print("Range Creation Examples:")
print(f"range(5): {list(range1)}, type: {type(range1)}")
print(f"range(2, 8): {list(range2)}")
print(f"range(0, 10, 2): {list(range3)}")
print(f"range(10, 0, -1): {list(range4)}")

# Range characteristics
print("\n" + "="*50)
print("Range Characteristics:")
print("="*50)

r = range(5, 15, 2)
print(f"\nRange: range(5, 15, 2)")
print(f"Length: {len(r)}")
print(f"First item: {r[0]}")
print(f"Last item: {r[-1]}")
print(f"All items: {list(r)}")
print(f"Is 7 in range? {7 in r}")
print(f"Is 8 in range? {8 in r}")

# Using range in loops
print("\n" + "="*50)
print("Using Range in Loops:")
print("="*50)

print("\nLoop with range(5):")
for i in range(5):
    print(f"  {i}", end=" ")
print()

print("\nLoop with range(1, 6):")
for i in range(1, 6):
    print(f"  {i}", end=" ")
print()

print("\nLoop with range(0, 10, 3):")
for i in range(0, 10, 3):
    print(f"  {i}", end=" ")
print()

print("\nReverse loop with range(5, 0, -1):")
for i in range(5, 0, -1):
    print(f"  {i}", end=" ")
print()

# Range with enumerate
print("\n" + "="*50)
print("Range with enumerate():")
print("="*50)

colors = ["red", "green", "blue", "yellow"]
print(f"\nList: {colors}")
print("Using enumerate with range:")
for i in range(len(colors)):
    print(f"  Index {i}: {colors[i]}")

# Convert range to different types
print("\n" + "="*50)
print("Converting Range to Other Types:")
print("="*50)

r = range(10, 20, 2)
print(f"\nOriginal range(10, 20, 2):")
print(f"As list: {list(r)}")
print(f"As tuple: {tuple(r)}")
print(f"As set: {set(r)}")

# Memory efficiency
print("\n" + "="*50)
print("Memory Efficiency of Range:")
print("="*50)

small_list = list(range(5))
small_range = range(5)
big_list = list(range(1000000))
big_range = range(1000000)

import sys
print(f"\nSmall range size: {sys.getsizeof(small_range)} bytes")
print(f"Small list size: {sys.getsizeof(small_list)} bytes")
print(f"\nBig range size: {sys.getsizeof(big_range)} bytes")
print(f"Big list size: {sys.getsizeof(big_list)} bytes")
print(f"\nRange is more memory efficient for large sequences!")
```

## Section 9: Type Conversion & Casting

### 9.1 Type Conversion Basics
- Converting one data type to another
- Implicit conversion: Python automatically converts
- Explicit conversion: Using constructors or functions
- Common functions: int(), float(), str(), list(), tuple(), dict(), set()

```python
# Section 9: Type Conversion Examples

# Implicit type conversion
print("Implicit Type Conversion:")
print("="*50)

x = 10        # int
y = 2.5       # float
result = x + y  # Python automatically converts int to float

print(f"x = {x} (type: {type(x).__name__})")
print(f"y = {y} (type: {type(y).__name__})")
print(f"x + y = {result} (type: {type(result).__name__})")

# Explicit type conversion - To Integer
print("\n" + "="*50)
print("Converting to Integer (int):")
print("="*50)

print(f"\nint('42'): {int('42')}")
print(f"int(3.14): {int(3.14)}")  # Truncates decimal
print(f"int(3.9): {int(3.9)}")    # Still truncates
print(f"int(True): {int(True)}")
print(f"int(False): {int(False)}")
print(f"int('0xFF', 16): {int('0xFF', 16)}")  # From hex string

# Error handling for invalid conversion
try:
    print(f"int('hello'): ", end="")
    print(int('hello'))
except ValueError as e:
    print(f"Error - Cannot convert 'hello' to int")

# Converting to Float
print("\n" + "="*50)
print("Converting to Float (float):")
print("="*50)

print(f"float('3.14'): {float('3.14')}")
print(f"float(42): {float(42)}")
print(f"float(True): {float(True)}")
print(f"float('inf'): {float('inf')}")
print(f"float('-inf'): {float('-inf')}")

# Converting to String
print("\n" + "="*50)
print("Converting to String (str):")
print("="*50)

print(f"str(42): '{str(42)}'")
print(f"str(3.14): '{str(3.14)}'")
print(f"str(True): '{str(True)}'")
print(f"str([1, 2, 3]): '{str([1, 2, 3])}'")

# Converting to Boolean
print("\n" + "="*50)
print("Converting to Boolean (bool):")
print("="*50)

print(f"bool(1): {bool(1)}")
print(f"bool(0): {bool(0)}")
print(f"bool('Hello'): {bool('Hello')}")
print(f"bool(''): {bool('')}")
print(f"bool([1, 2]): {bool([1, 2])}")
print(f"bool([]): {bool([])}")
print(f"bool(None): {bool(None)}")

# Converting to List
print("\n" + "="*50)
print("Converting to List (list):")
print("="*50)

print(f"list((1, 2, 3)): {list((1, 2, 3))}")
print(f"list('ABC'): {list('ABC')}")
print(f"list(range(5)): {list(range(5))}")
print(f"list({{1, 2, 3}}): {list({1, 2, 3})}")

# Converting to Tuple
print("\n" + "="*50)
print("Converting to Tuple (tuple):")
print("="*50)

print(f"tuple([1, 2, 3]): {tuple([1, 2, 3])}")
print(f"tuple('ABC'): {tuple('ABC')}")
print(f"tuple(range(5)): {tuple(range(5))}")
print(f"tuple({{1, 2, 3}}): {tuple({1, 2, 3})}")

# Converting to Set
print("\n" + "="*50)
print("Converting to Set (set):")
print("="*50)

print(f"set([1, 2, 2, 3, 3]): {set([1, 2, 2, 3, 3])}")
print(f"set('AAB'): {set('AAB')}")
print(f"set(range(5)): {set(range(5))}")

# Checking data types
print("\n" + "="*50)
print("Checking Data Types with type():")
print("="*50)

values = [42, 3.14, "hello", True, [1, 2], {1, 2}, {"a": 1}, None]

for val in values:
    print(f"type({repr(val)}): {type(val)}")

# Using isinstance() for type checking
print("\n" + "="*50)
print("Using isinstance() for Type Checking:")
print("="*50)

x = 42
print(f"\nx = {x}")
print(f"isinstance(x, int): {isinstance(x, int)}")
print(f"isinstance(x, float): {isinstance(x, float)}")
print(f"isinstance(x, (int, float)): {isinstance(x, (int, float))}")
```

## Section 10: NoneType Data Type

### 10.1 None Basics
- Represents absence of a value
- Only one value: None
- Different from 0, False, or empty string
- Often used as default values
- Used when function doesn't return anything

```python
# Section 10: NoneType Examples

# Creating None values
empty_value = None
result = None

print("NoneType Examples:")
print(f"empty_value = {empty_value}, type: {type(empty_value)}")
print(f"result = {result}, type: {type(result)}")

# Using None as default value
print("\n" + "="*50)
print("None as Default Value:")
print("="*50)

# Function without return statement returns None
def greet(name):
    print(f"Hello, {name}!")
    # No return statement

result = greet("Alice")
print(f"\nFunction without return: result = {result}")

# Function with conditional return
def find_value(items, target):
    for item in items:
        if item == target:
            return item
    return None  # Not found

print(f"find_value([1, 2, 3], 2): {find_value([1, 2, 3], 2)}")
print(f"find_value([1, 2, 3], 5): {find_value([1, 2, 3], 5)}")

# Checking for None
print("\n" + "="*50)
print("Checking for None:")
print("="*50)

x = None
y = 0
z = ""

print(f"\nx = {x}")
print(f"x is None: {x is None}")
print(f"x == None: {x == None}")

print(f"\ny = {y}")
print(f"y is None: {y is None}")
print(f"y == 0: {y == 0}")

print(f"\nz = '{z}'")
print(f"z is None: {z is None}")
print(f"z == '': {z == ''}")

# None vs Falsy values
print("\n" + "="*50)
print("None vs Other Falsy Values:")
print("="*50)

falsy_values = [None, 0, 0.0, False, "", [], {}, set()]
print("\nFalsy values in Python:")
for val in falsy_values:
    print(f"  {repr(val):15} -> bool({repr(val)}) = {bool(val)}")

# Using None in data structures
print("\n" + "="*50)
print("None in Data Structures:")
print("="*50)

# List with None
mixed_list = [1, None, "hello", None, 3.14]
print(f"\nList: {mixed_list}")
print(f"Count of None: {mixed_list.count(None)}")
print(f"Index of first None: {mixed_list.index(None)}")

# Dictionary with None values
data = {
    "name": "John",
    "age": None,
    "email": "john@example.com",
    "phone": None
}
print(f"\nDictionary: {data}")
for key, value in data.items():
    if value is None:
        print(f"  {key}: Not specified")
    else:
        print(f"  {key}: {value}")

# Handling None safely
print("\n" + "="*50)
print("Safe Handling of None Values:")
print("="*50)

user_data = {"name": "Alice", "age": None}

# Using get() with default
age = user_data.get("age", "Not provided")
print(f"\nAge (using get with default): {age}")

# Using if-else
if user_data["age"] is None:
    age_display = "Age not provided"
else:
    age_display = f"Age: {user_data['age']}"
print(f"Age (using if-else): {age_display}")

# Using ternary operator
city = user_data.get("city")
city_display = user_data.get("city") if user_data.get("city") is not None else "Location unknown"
print(f"City: {city_display}")
```
